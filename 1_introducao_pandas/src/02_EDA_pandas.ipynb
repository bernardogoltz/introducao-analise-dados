## Visualização

```python
import matplotlib.pyplot as plt
print(df.columns)
```

```python
df_grouped = df.groupby(
    ['SigAgenteDistribuidora','DscSubGrupoTarifario','DatCompetencia']
)['Receita Energia (R$)'].sum().reset_index()

df_grouped
```

```python
df_geral = df_grouped.groupby('DatCompetencia')['Receita Energia (R$)'].sum().reset_index()


#help(pd.to_datetime)

setup_dt = {
    'format':'%Y-%m-%d'
}

df_geral['DatCompetencia'] = pd.to_datetime(
    df_geral['DatCompetencia'],
    **setup_dt
).dt.strftime('%m')
```

```python
import matplotlib.pyplot as plt

figure, ax = plt.subplots(
    figsize=(8, 4),
    dpi=150
)

ax.plot(
    df_geral['DatCompetencia'],
    df_geral['Receita Energia (R$)'],
    color = 'black',
    linewidth = 2
)

ax.set_ylim(
    0,
    df_geral['Receita Energia (R$)'].max() * 1.1
)

ax.set_ylabel('Bilhões de R$')
ax.set_xlabel('Mês')

plt.grid(True, alpha = 0.2)
plt.title('Receita anual Distribuidoras SIN')
plt.show()
```

```python
#df.groupby()
colunas = ['DscClasseConsumoMercado','DatCompetencia']
variaveis = ['Receita Energia (R$)', 'Energia TUSD (kWh)','Energia TE (kWh)']

df_grouped_classes = df.groupby(colunas)[variaveis].sum().reset_index()

df_grouped_classes['DatCompetencia'] = pd.to_datetime(
    df_grouped_classes['DatCompetencia'],
    **setup_dt
).dt.strftime('%m')

df_grouped_classes
```

```python
fig, ax = plt.subplots(figsize=(15, 5), dpi=150)

classes = ['Comercial','Residencial','Rural']

for classe in ['Comercial','Residencial','Rural']:
    subset = df_grouped_classes[
        df_grouped_classes['DscClasseConsumoMercado'] == classe
    ]

    ax.plot(
        subset['DatCompetencia'],
        subset['Receita Energia (R$)']/1e9,
        label=classe,
        linewidth = 2
    )

    
plt.title("Receita Anual por classe de consumo")
plt.ylabel("Bilhões R$")

plt.legend(title = 'Classe Consumo')
plt.grid(alpha = 0.2)
#plt.yscale('symlog')
```

```python

colunas = ['SigAgenteDistribuidora','DscClasseConsumoMercado']

df_ucs = df.groupby(colunas)['Número de consumidores'].sum().reset_index()

df_ucs['DscClasseConsumoMercado'].unique()

classes_principais = ['Comercial', 'Residencial', 'Rural']

df_ucs['DscClasseConsumoMercado'] = (
    df_ucs['DscClasseConsumoMercado']
    .mask(
        ~df_ucs['DscClasseConsumoMercado'].isin(classes_principais),
        'Outros'
    )
)

df_ucs = (
    df_ucs.groupby(colunas)['Número de consumidores']
    .sum()
    .reset_index()
)
```

```python
top_10 = list(df_ucs.groupby('SigAgenteDistribuidora')['Número de consumidores'].sum().nlargest(4).index)

df_top_4 = df_ucs[
    df_ucs['SigAgenteDistribuidora'].isin(top_10)
]

len(df_top_4)
df_top_4
```

```python
import numpy as np
import matplotlib.pyplot as plt

figure, ax = plt.subplots(
    figsize=(18, 7)
)

classes = ['Residencial', 'Comercial', 'Rural', 'Outros']

distribuidoras = (
    df_top_4
    .groupby('SigAgenteDistribuidora')['Número de consumidores']
    .sum()
    .sort_values(ascending=False)
    .index
)

x = np.arange(len(distribuidoras))
largura = 0.2

for i, classe in enumerate(classes):

    subset = (
        df_top_4[
            df_top_4['DscClasseConsumoMercado'] == classe
        ]
        .set_index('SigAgenteDistribuidora')
        .reindex(distribuidoras)
    )

    ax.bar(
        x + i * largura,
        subset['Número de consumidores'],
        width=largura,
        label=classe,
        edgecolor='black',
        linewidth=0.8,
        zorder = 1
    )

ax.set_xticks(
    x + largura * (len(classes) - 1) / 2,
    distribuidoras
)

ax.set_yscale('log')

ax.grid(
    axis='y',
    alpha=0.3,
    zorder = 0
)

totais = (
    df_top_4
    .groupby('SigAgenteDistribuidora')['Número de consumidores']
    .sum()
    .reindex(distribuidoras)
)

centro_grupo = x + largura * (len(classes) - 1) / 2

for xpos, distribuidora in zip(centro_grupo, distribuidoras):

    total = totais.loc[distribuidora]

    maior_barra = (
        df_top_4[
            df_top_4['SigAgenteDistribuidora'] == distribuidora
        ]['Número de consumidores']
        .max()
    )

    ax.annotate(
        f'{total / 1_000_000:.2f}M',
        xy=(xpos, maior_barra),
        xytext=(0, 6),
        textcoords='offset points',
        ha='center',
        va='bottom',
    )

ax.legend()
ax.set_axisbelow(True)
plt.show()
```
